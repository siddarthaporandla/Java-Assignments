A vampire number v is a number with an even number of digits n, that can be factored into two numbers x and y each with n/2 digits and not both with trailing zeroes, where v contains precisely all the digits from x and from y, in any order. Write a java program to print first 100 vampire numbers.

```java
// First 100 Vampire Numbers
static boolean isVampire(int number) {
    String value = String.valueOf(number);
    int digits = value.length();

    if (digits % 2 != 0) {
        return false;
    }

    int half = digits / 2;
    int min = (int) Math.pow(10, half - 1);
    int max = (int) Math.pow(10, half) - 1;

    for (int x = min; x <= max; x++) {
        if (number % x == 0) {
            int y = number / x;

            if (y < min || y > max) {
                continue;
            }

            if (x % 10 == 0 && y % 10 == 0) {
                continue;
            }

            String combined = String.valueOf(x) + String.valueOf(y);

            char[] original = value.toCharArray();
            char[] factors = combined.toCharArray();

            java.util.Arrays.sort(original);
            java.util.Arrays.sort(factors);

            if (java.util.Arrays.equals(original, factors)) {
                return true;
            }
        }
    }

    return false;
}

int count = 0;
int number = 1;

while (count < 100) {
    if (isVampire(number)) {
        System.out.print(number + " ");
        count++;
    }

    number++;
}
```

```java
// Constructor Overloading & Constructor Chaining
class Student {
    String name;
    int age;

    Student() {
        this("Siddu", 20);
        System.out.println("First constructor");
    }

    Student(String name, int age) {
        this.name = name;
        this.age = age;
        System.out.println("Second constructor");
    }

    void display() {
        System.out.println("Name: " + name);
        System.out.println("Age: " + age);
    }
}

Student student = new Student();
student.display();
```
