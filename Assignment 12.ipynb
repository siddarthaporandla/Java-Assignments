```java
import java.util.*;
import java.util.stream.Collectors;

// Student Class
class Student {
    private int id;
    private String name;
    private int age;
    private String gender;
    private String engDepartment;
    private int yearOfEnrollment;
    private double perTillDate;

    public Student(int id, String name, int age, String gender, String engDepartment, int yearOfEnrollment,      double perTillDate) {
        this.id = id;
        this.name = name;
        this.age = age;
        this.gender = gender;
        this.engDepartment = engDepartment;
        this.yearOfEnrollment = yearOfEnrollment;
        this.perTillDate = perTillDate;
    }

    public int getId() {
        return id;
    }

    public String getName() {
        return name;
    }

    public int getAge() {
        return age;
    }

    public String getGender() {
        return gender;
    }

    public String getEngDepartment() {
        return engDepartment;
    }

    public int getYearOfEnrollment() {
        return yearOfEnrollment;
    }

    public double getPerTillDate() {
        return perTillDate;
    }

    @Override
    public String toString() {
        return "Student{id=" + id + ", name='" + name + "', age=" + age +
               ", gender='" + gender + "', department='" + engDepartment +
               "', year=" + yearOfEnrollment + ", percentage=" + perTillDate + "}";
    }
}


// Stream API Operations

class StudentAnalytics {

    // 1. Print names of all departments
    public static void printDepartments(List<Student> students) {
        System.out.println("1. Departments:");

        students.stream()
                .map(Student::getEngDepartment)
                .distinct()
                .forEach(System.out::println);
    }


    // 2. Students enrolled after 2018
    public static void studentsEnrolledAfter2018(List<Student> students) {
        System.out.println("\n2. Students enrolled after 2018:");

        students.stream()
                .filter(student -> student.getYearOfEnrollment() > 2018)
                .map(Student::getName)
                .forEach(System.out::println);
    }


    // 3. Male students in Computer Science
    public static void maleComputerScienceStudents(List<Student> students) {
        System.out.println("\n3. Male students in Computer Science:");

        students.stream()
                .filter(student -> student.getGender().equals("Male"))
                .filter(student -> student.getEngDepartment().equals("Computer Science"))
                .forEach(System.out::println);
    }


    // 4. Number of male and female students
    public static void countByGender(List<Student> students) {
        System.out.println("\n4. Number of male and female students:");

        Map<String, Long> result = students.stream()
                .collect(Collectors.groupingBy(
                        Student::getGender,
                        Collectors.counting()
                ));

        System.out.println(result);
    }


    // 5. Average age of male and female students
    public static void averageAgeByGender(List<Student> students) {
        System.out.println("\n5. Average age by gender:");

        Map<String, Double> result = students.stream()
                .collect(Collectors.groupingBy(
                        Student::getGender,
                        Collectors.averagingInt(Student::getAge)
                ));

        System.out.println(result);
    }


    // 6. Student with highest percentage
    public static void highestPercentageStudent(List<Student> students) {
        System.out.println("\n6. Student with highest percentage:");

        Student result = students.stream()
                .max(Comparator.comparingDouble(Student::getPerTillDate))
                .orElse(null);

        System.out.println(result);
    }


    // 7. Number of students in each department
    public static void countByDepartment(List<Student> students) {
        System.out.println("\n7. Number of students in each department:");

        Map<String, Long> result = students.stream()
                .collect(Collectors.groupingBy(
                        Student::getEngDepartment,
                        Collectors.counting()
                ));

        System.out.println(result);
    }


    // 8. Average percentage in each department
    public static void averagePercentageByDepartment(List<Student> students) {
        System.out.println("\n8. Average percentage in each department:");

        Map<String, Double> result = students.stream()
                .collect(Collectors.groupingBy(
                        Student::getEngDepartment,
                        Collectors.averagingDouble(Student::getPerTillDate)
                ));

        result.forEach((department, average) ->
                System.out.println(department + " -> " + average)
        );
    }


    // 9. Youngest male student in Electronic department
    public static void youngestMaleElectronicStudent(List<Student> students) {
        System.out.println("\n9. Youngest male student in Electronic department:");

        Optional<Student> result = students.stream()
                .filter(student -> student.getGender().equals("Male"))
                .filter(student -> student.getEngDepartment().equals("Electronic"))
                .collect(Collectors.minBy(
                        Comparator.comparingInt(Student::getAge)
                ));

        result.ifPresent(System.out::println);
    }


    // 10. Male and female students in Computer Science
    public static void countComputerScienceByGender(List<Student> students) {
        System.out.println("\n10. Male and female students in Computer Science:");

        Map<String, Long> result = students.stream()
                .filter(student -> student.getEngDepartment().equals("Computer Science"))
                .collect(Collectors.groupingBy(
                        Student::getGender,
                        Collectors.counting()
                ));

        System.out.println(result);
    }
}


// Student Data

List<Student> students = Arrays.asList(
        new Student(111, "Jiya Brein", 17, "Female", "Computer Science", 2018, 70.8),
        new Student(122, "Paul Niksui", 18, "Male", "Mechanical", 2016, 50.2),
        new Student(133, "Martin Theron", 17, "Male", "Electronic", 2017, 90.3),
        new Student(144, "Murali Gowda", 18, "Male", "Electrical", 2018, 80.0),
        new Student(155, "Nima Roy", 19, "Female", "Textile", 2016, 70.0),
        new Student(166, "Iqbal Hussain", 18, "Male", "Security", 2016, 70.0),
        new Student(177, "Manu Sharma", 16, "Male", "Chemical", 2018, 70.0),
        new Student(188, "Wang Liu", 20, "Male", "Computer Science", 2015, 80.0),
        new Student(199, "Amelia Zoe", 18, "Female", "Computer Science", 2016, 85.0),
        new Student(200, "Jaden Dough", 18, "Male", "Security", 2015, 82.0),
        new Student(211, "Jasna Kaur", 20, "Female", "Electronic", 2019, 83.0),
        new Student(222, "Nitin Joshi", 19, "Male", "Textile", 2016, 60.4),
        new Student(233, "Jyothi Reddy", 16, "Female", "Computer Science", 2015, 45.6),
        new Student(244, "Nicolus Den", 16, "Male", "Electronic", 2017, 95.8),
        new Student(255, "Ali Baig", 17, "Male", "Electronic", 2018, 88.4),
        new Student(266, "Sanvi Pandey", 17, "Female", "Electric", 2019, 72.4),
        new Student(277, "Anuj Chettiar", 18, "Male", "Computer Science", 2017, 57.5)
);

// Run All Questions

StudentAnalytics.printDepartments(students);
StudentAnalytics.studentsEnrolledAfter2018(students);
StudentAnalytics.maleComputerScienceStudents(students);
StudentAnalytics.countByGender(students);
StudentAnalytics.averageAgeByGender(students);
StudentAnalytics.highestPercentageStudent(students);
StudentAnalytics.countByDepartment(students);
StudentAnalytics.averagePercentageByDepartment(students);
StudentAnalytics.youngestMaleElectronicStudent(students);
StudentAnalytics.countComputerScienceByGender(students);
```
